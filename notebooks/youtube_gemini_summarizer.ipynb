```python
# ✅ Imports
from google import genai
import pandas as pd
import time
```

```python
# ✅ Configure Gemini Client
client = genai.Client(api_key="AIzaSyAfd7Va3xPARxLeCHWm2YFHhAYMDIPhy1g")
```

```python
# ✅ Load video links
LINKS_PATH = "/content/drive/MyDrive/super-bowl-ad-analysis/data/raw/youtube_gemini/homepage_youtube_links.txt"
with open(LINKS_PATH, "r", encoding="utf-8") as f:
    video_pages = [line.strip() for line in f if line.strip()]
print(f"🔗 Loaded {len(video_pages)} video links")
```

```python
# ✅ Summarize using Gemini (with improved parsing)
summaries = []

for i, url in enumerate(video_pages):
    prompt = f"""
    Here's a video link: {url}

Please do the following:
1. Categorize brand name and year
2. Provide a concise summary of this Super Bowl commercial in **less than 100 words**. Do **not** start with phrases like "Here's a summary..." or "This is a summary...". Just provide the content directly.
3. Give a sentiment analysis using **2 to 3 adjectives only**. Do **not** include phrases like "Sentiment analysis:"—only list the adjectives.

Return your output in the following format:

Brand: <brand name>

Year: <year>

Summary: <your summary here>

Sentiment: <adjective1>, <adjective2>, <optional adjective3>
    """

    try:
        response = client.models.generate_content(
            model="gemini-2.0-flash",
            contents=prompt
        )

        output = response.text.strip()
        lines = output.split("\n")

        # Robust label-based parsing
        parsed = {"brand": "", "year": "", "summary": "", "sentiment": ""}

        for line in lines:
            if line.startswith("Brand:"):
                parsed["brand"] = line.replace("Brand:", "").strip()
            elif line.startswith("Year:"):
                parsed["year"] = line.replace("Year:", "").strip()
            elif line.startswith("Summary:"):
                parsed["summary"] = line.replace("Summary:", "").strip()
            elif line.startswith("Sentiment:"):
                parsed["sentiment"] = line.replace("Sentiment:", "").strip()

        summaries.append({
            "source_url": url,
            "brand": parsed["brand"],
            "year": parsed["year"],
            "summary": parsed["summary"],
            "sentiment": parsed["sentiment"]
        })

        print(f"✅ {i+1}/{len(video_pages)} done")
        time.sleep(5)

    except Exception as e:
        print(f"⚠️ Error on {url}: {e}")
```

```python

# ✅ Save to CSV
OUTPUT_CSV = "/content/drive/MyDrive/super-bowl-ad-analysis/data/raw/youtube_gemini/video_summaries_sentiment.csv"
df = pd.DataFrame(summaries)
df.to_csv(OUTPUT_CSV, index=False)
print(f"\n✅ Saved {len(df)} entries to: {OUTPUT_CSV}")
```
